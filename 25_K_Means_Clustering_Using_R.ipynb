```r
install.packages("readxl")
library(readxl)

myData <- read_excel("/content/CandyBars.xlsx")
myData
```

Install and load the cluster package

```r
install.packages("cluster")

library(cluster)
```

```r
# exclude the Brands variable from the analysis and standardize the other variables using the scale function

myData_std <- scale(myData[ ,2:5])
```

We use the **pam** function because it is considered more
robust than other similar functions

The acronym pam stands for Partitioning Around Medoids

```r
# we set the option k to 4 because we have preselected 4 clusters

set.seed(1)
kResult <- pam(myData_std, k = 4)
summary(kResult)
```

The R results present information on medoids
instead of centroids

Medoids are similar to centroids, but they
are observations of the data set

A medoid of a cluster is the
observation whose average distance to all other observations in
the same cluster is the shortest

So, a medoid is the observation that is closest to the center of the cluster; therefore, the values of its variables are often used to represent the cluster averages or the values for the cluster center

Because we cannot plot a fourdimensional
graph based on the four variables, we use PCA in
order to plot the clusters onto a two-dimensional graph with the
first two principal components as the x- and y-axes

```r
plot(kResult)
```
